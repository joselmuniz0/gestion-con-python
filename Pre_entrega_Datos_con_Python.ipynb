## Actividad 2: Cálculo de ventas mensuales
Se calcula el total de ventas por mes usando variables
y operadores aritméticos en Python.

```python
# Bloque de CÓDIGO — Ventas mensuales

# --- PASO 1: Limpiar la columna precio ---
# precio tiene formato "$69.94" (texto). Hay que eliminar el "$"
# y convertir a número decimal (float)

df_ventas['precio_num'] = (
    df_ventas['precio']
    .str.replace('$', '', regex=False)
    .astype(float)
)

# --- PASO 2: Calcular el ingreso por fila ---
# ingreso = precio * cantidad  (operador aritmético *)
df_ventas['ingreso'] = df_ventas['precio_num'] * df_ventas['cantidad']

# --- PASO 3: Convertir fecha a tipo datetime para extraer el mes ---
df_ventas['fecha_venta'] = pd.to_datetime(df_ventas['fecha_venta'],
                                              format='%d/%m/%Y')
df_ventas['mes'] = df_ventas['fecha_venta'].dt.month

# --- PASO 4: Agrupar por mes y sumar ingresos ---
ventas_mensuales = df_ventas.groupby('mes')['ingreso'].sum()

# --- PASO 5: Mostrar resultados con variables ---
total_anual   = ventas_mensuales.sum()
promedio_mes  = ventas_mensuales.mean()
mes_mayor     = ventas_mensuales.idxmax()
ingreso_mayor = ventas_mensuales.max()

print("=== Ventas Mensuales ===")
print(ventas_mensuales.round(2))
print()
print(f"Total anual:         ${total_anual:,.2f}")
print(f"Promedio mensual:    ${promedio_mes:,.2f}")
print(f"Mes con más ventas:  Mes {mes_mayor} (${ingreso_mayor:,.2f})")
```

# 3 - Almacenar datos de ventas: ¿diccionario o lista?
Elegir la estructura correcta y justificar la decisión
Fundamento teórico *texto en cursiva*



La estructura correcta es la de Diccionario, ya que nos proporciona
acceso por clave con nombre ('producto', 'precio'). Código legible. Sin ambigüedad.

```python
# Bloque de CÓDIGO — Estructuras de datos para ventas

# --- OPCIÓN A: Lista (menos recomendada) ---
# El orden importa: [producto, precio, cantidad]
# Si alguien cambia el orden, el programa falla
venta_lista = ["Lámpara de mesa", 105.10, 5]
print("Lista - Producto:", venta_lista[0])  # acceso por índice
print("Lista - Precio:  ", venta_lista[1])

print()

# --- OPCIÓN B: Diccionario (recomendada) ---
# Acceso por clave con nombre: no hay ambigüedad
venta_dict = {
    "producto":  "Lámpara de mesa",
    "precio":    105.10,
    "cantidad":  5
}
print("Dict - Producto:", venta_dict["producto"])
print("Dict - Precio:  ", venta_dict["precio"])
print("Dict - Ingreso: ", venta_dict["precio"] * venta_dict["cantidad"])

print()

# --- OPCIÓN C: Lista de diccionarios (para múltiples ventas) ---
# Esta es la estructura más práctica para varios registros
registro_ventas = [
    {"producto": "Lámpara de mesa",   "precio": 105.10, "cantidad": 5},
    {"producto": "Cuadro decorativo",  "precio": 69.94,  "cantidad": 3},
    {"producto": "Heladera",            "precio": 114.35, "cantidad": 8},
]

print("=== Registro de ventas ===")
for venta in registro_ventas:
    ingreso = venta["precio"] * venta["cantidad"]
    print(f"  {venta['producto']:<25} → ${ingreso:.2f}")

# --- CONCLUSIÓN: el diccionario es mejor porque ---
# 1) Las claves tienen nombre (legibilidad)
# 2) No depende del orden de los elementos
# 3) Es la base de un DataFrame de Pandas
```

# 5 - Identificar valores nulos y duplicados en los tres datasets

Documentar el estado inicial de los datos (sin limpiar todavía)

**¿Por qué es importante?**

Los datos reales siempre tienen problemas. Identificarlos antes de cualquier análisis es obligatorio: un promedio calculado sobre datos con nulos silenciosos produce resultados incorrectos. Identificamos y documentamos.



```python
# Bloque de CÓDIGO — Identificación de valores nulos

print("=" * 50)
print("VALORES NULOS POR DATASET Y COLUMNA")
print("=" * 50)

for nombre, df in [("VENTAS", df_ventas),
                    ("CLIENTES", df_clientes),
                    ("MARKETING", df_marketing)]:
    nulos = df.isnull().sum()
    total = df.isnull().sum().sum()
    print(f"\n--- {nombre} (total nulos: {total}) ---")
    print(nulos[nulos > 0] if total > 0 else "  Sin valores nulos.")
```

```python
# Bloque de CÓDIGO — Ver las filas con nulos en ventas

print("=== Filas con valores nulos en df_ventas ===")
filas_nulas = df_ventas[df_ventas.isnull().any(axis=1)]
print(filas_nulas)

# axis=1 significa "revisar a lo largo de las columnas"
# any() devuelve True si al menos un valor en esa fila es nulo
```
