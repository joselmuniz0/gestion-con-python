## 4. Actividad 1.2 — Cálculo de ventas mensuales

**Qué se hace:** calcular el total facturado por mes a partir del dataset `df_ventas`.

**Por qué se hace en dos versiones:**
- La **versión didáctica con Python puro** demuestra el manejo de variables, operadores y bucles (Clase 1).
- La **versión con Pandas** muestra cómo se resuelve el mismo problema con la herramienta apropiada.

**Detalle técnico:** la columna `precio` viene con `$` adelante (string). En esta sección lo limpiamos *localmente* sobre una copia, sin tocar `df_ventas`. La limpieza definitiva se hace en la Etapa 2.

### 4.1 — Versión didáctica (Python puro)

```python
# 4.1 — Cálculo de ventas mensuales con Python puro

ventas_demo = [
    {'fecha': '02/01/2024', 'producto': 'Cuadro decorativo', 'precio': 69.94, 'cantidad': 5},
    {'fecha': '15/01/2024', 'producto': 'Lámpara de mesa',   'precio': 105.10, 'cantidad': 3},
    {'fecha': '03/02/2024', 'producto': 'Heladera',          'precio': 850.00, 'cantidad': 2},
    {'fecha': '20/02/2024', 'producto': 'Tablet',            'precio': 320.00, 'cantidad': 4},
    {'fecha': '10/03/2024', 'producto': 'Smartphone',        'precio': 600.00, 'cantidad': 6},
]

ventas_por_mes = {}
for registro in ventas_demo:
    dia, mes, anio = registro['fecha'].split('/')
    clave_mes = f'{anio}-{mes}'
    importe = registro['precio'] * registro['cantidad']
    ventas_por_mes[clave_mes] = ventas_por_mes.get(clave_mes, 0) + importe

print('Ventas mensuales (versión didáctica con Python puro):')
print('-' * 40)
for mes in sorted(ventas_por_mes):
    print(f'  {mes}:  $ {ventas_por_mes[mes]:>12,.2f}')
```

### 4.2 — Versión con Pandas (sobre los datos reales)

```python
# 4.2 — Cálculo de ventas mensuales con Pandas (sobre copia local)

# 1) Copia local: cualquier transformación NO afecta a df_ventas.
ventas_calc = df_ventas.copy()

# 2) Limpieza local del precio: sacar '$' y convertir a float.
ventas_calc['precio'] = pd.to_numeric(
    ventas_calc['precio'].str.replace('$', '', regex=False),
    errors='coerce'
)

# 3) Conversión local de la fecha.
ventas_calc['fecha_venta'] = pd.to_datetime(
    ventas_calc['fecha_venta'], dayfirst=True, errors='coerce'
)

# 4) Cálculo del importe línea por línea.
ventas_calc['importe'] = ventas_calc['precio'] * ventas_calc['cantidad']

# 5) Agrupamiento por mes con dt.to_period('M') (preserva orden cronológico).
ventas_mensuales = (
    ventas_calc
    .groupby(ventas_calc['fecha_venta'].dt.to_period('M'))['importe']
    .sum()
    .sort_index()
)

print('Ventas mensuales reales (Pandas):')
print('-' * 40)
print(ventas_mensuales.apply(lambda x: f'$ {x:>14,.2f}').to_string())
print('-' * 40)
print(f'TOTAL del período: $ {ventas_mensuales.sum():,.2f}')
```

## 5. Actividad 1.3 — Estructuras de datos para almacenar ventas

**Decisión:** lista de diccionarios.

**Razones:**
1. Cada venta tiene campos heterogéneos con nombre → el diccionario es la representación natural.
2. La lista preserva el orden cronológico y admite duplicados legítimos.
3. Es directamente convertible a `pandas.DataFrame` con `pd.DataFrame(lista)`.

```python
# 5 — Almacenamiento de ventas con lista de diccionarios

ventas = [
    {'producto': 'Cuadro decorativo', 'precio':  69.94, 'cantidad': 5},
    {'producto': 'Lámpara de mesa',   'precio': 105.10, 'cantidad': 3},
    {'producto': 'Heladera',          'precio': 850.00, 'cantidad': 2},
    {'producto': 'Tablet',            'precio': 320.00, 'cantidad': 4},
    {'producto': 'Smartphone',        'precio': 600.00, 'cantidad': 6},
    {'producto': 'Heladera',          'precio': 850.00, 'cantidad': 1},
]

print('Detalle línea por línea:')
print('-' * 60)
for v in ventas:
    importe = v['precio'] * v['cantidad']
    print(f"  {v['producto']:<20}  ${v['precio']:>7.2f} x {v['cantidad']}  =  ${importe:>10,.2f}")

total = sum(v['precio'] * v['cantidad'] for v in ventas)
print('-' * 60)
print(f'Total facturado: $ {total:,.2f}\n')

totales_por_producto = {}
for v in ventas:
    p = v['producto']
    totales_por_producto[p] = totales_por_producto.get(p, 0) + v['precio'] * v['cantidad']

print('Total por producto:')
print('-' * 40)
for p, t in sorted(totales_por_producto.items(), key=lambda x: -x[1]):
    print(f'  {p:<20}  $ {t:>10,.2f}')

df_ventas_demo = pd.DataFrame(ventas)
print('\nMismo conjunto como DataFrame:')
df_ventas_demo
```

## 6. Actividad 1.4 — Análisis exploratorio inicial (EDA)

Encapsulamos la exploración en una función reutilizable para garantizar criterio uniforme entre los 3 datasets.

```python
# 6.1 — Función reutilizable de EDA

def explorar_dataframe(df: pd.DataFrame, nombre: str) -> None:
    '''Imprime un resumen exploratorio estándar del DataFrame.'''
    sep = '=' * 70
    print(sep)
    print(f'  EDA — {nombre}')
    print(sep)

    print(f'\nForma: {df.shape[0]} filas x {df.shape[1]} columnas')

    print('\nTipos de datos y memoria:')
    df.info(memory_usage='deep')

    print('\nPrimeras 5 filas:')
    print(df.head())
    print('\nÚltimas 5 filas:')
    print(df.tail())

    print('\nEstadísticos numéricos:')
    print(df.describe())

    cols_obj = df.select_dtypes(include=['object', 'category']).columns
    if len(cols_obj) > 0:
        print('\nEstadísticos de columnas categóricas:')
        print(df[cols_obj].describe())

    print('\nValores únicos por columna:')
    print(df.nunique().sort_values(ascending=False))

    print('\n' + sep + '\n')
```

## 7. Actividad 1.5 — Calidad de datos: nulos y duplicados

Identificación (no corrección) de valores nulos y registros duplicados, con informe consolidado del estado inicial.

```python
# 7.1 — Función de informe de nulos por columna

def informe_nulos(df: pd.DataFrame, nombre: str) -> pd.DataFrame:
    cantidad   = df.isnull().sum()
    porcentaje = (cantidad / len(df) * 100).round(2)
    informe = (
        pd.DataFrame({
            'columna': cantidad.index,
            'nulos': cantidad.values,
            '% nulos': porcentaje.values,
        })
        .sort_values('nulos', ascending=False)
        .reset_index(drop=True)
    )
    print(f'\n--- Nulos en {nombre} ---')
    print(f'Total de nulos en el dataset: {int(df.isnull().sum().sum())}')
    print(informe.to_string(index=False))
    return informe

nulos_clientes  = informe_nulos(df_clientes,  'Clientes')
nulos_ventas    = informe_nulos(df_ventas,    'Ventas')
nulos_marketing = informe_nulos(df_marketing, 'Marketing')
```

```python
# 7.2 — Función de informe de duplicados

def informe_duplicados(df: pd.DataFrame, nombre: str, subset=None) -> pd.DataFrame:
    mask = df.duplicated(subset=subset, keep=False)
    duplicados = df[mask].sort_values(by=subset or list(df.columns))

    print(f'\n--- Duplicados en {nombre} ---')
    print(f'Filas totales: {len(df)}')
    print(f'Filas duplicadas (todas las apariciones): {len(duplicados)}')
    print(f'Filas únicas si se eliminaran duplicados: '
          f'{len(df) - int(df.duplicated(subset=subset).sum())}')
    return duplicados

dup_clientes  = informe_duplicados(df_clientes,  'Clientes')
dup_ventas    = informe_duplicados(df_ventas,    'Ventas')
dup_marketing = informe_duplicados(df_marketing, 'Marketing')
```

```python
# 7.3 — Informe consolidado del estado inicial (consigna)

def informe_inicial(df: pd.DataFrame, nombre: str) -> dict:
    print(f'\n{"#"*60}')
    print(f'  INFORME INICIAL — {nombre}')
    print(f'{"#"*60}')

    total_registros = len(df)
    nulos_por_col   = df.isnull().sum()
    total_nulos     = int(nulos_por_col.sum())
    duplicados      = int(df.duplicated().sum())

    print(f'\na) Cantidad total de registros: {total_registros}')
    print(f'\nb) Valores nulos por columna:')
    if total_nulos > 0:
        print(nulos_por_col[nulos_por_col > 0].to_string())
    else:
        print('   (no hay nulos)')
    print(f'\n   Total de nulos en el dataset: {total_nulos}')
    print(f'\nc) Filas duplicadas (criterio: todas las columnas): {duplicados}')

    return {
        'dataset': nombre, 'registros': total_registros,
        'nulos_total': total_nulos,
        'nulos_por_columna': nulos_por_col.to_dict(),
        'duplicados_total': duplicados,
    }

resumen_clientes  = informe_inicial(df_clientes,  'Clientes')
resumen_ventas    = informe_inicial(df_ventas,    'Ventas')
resumen_marketing = informe_inicial(df_marketing, 'Marketing')
```

# ETAPA 2 — Preprocesamiento y Limpieza de Datos


## 8. Actividad 2.1 — Limpieza de datos (Clase 5)

**Qué se hace:** corregir los problemas detectados en la Etapa 1 produciendo DataFrames limpios.

**Decisiones tomadas y fundamento:**

| Problema detectado | Decisión | Fundamento |
|---|---|---|
| 35 filas duplicadas en `df_ventas` | Eliminar conservando la primera ocurrencia | Son filas íntegramente repetidas (incluso `id_venta`); inflarían las métricas. |
| 2 filas con nulos en `precio` y `cantidad` simultáneamente | Eliminarlas | Sin precio ni cantidad la fila es inutilizable; representa el 0.07 % del dataset (impacto despreciable). |
| `precio` con símbolo `$` | Sacar `$` y convertir a `float` | Sin esto, no se puede operar matemáticamente con la columna. |
| Fechas como string en `dd/mm/yyyy` | Convertir a `datetime` con `dayfirst=True` | Permite ordenar cronológicamente, agrupar por mes/año y calcular diferencias. |
| Categorías y canales como string | Mantener como `object`/`category` | No requieren limpieza; los valores son consistentes (verificado en EDA). |

**Principio rector:** los DataFrames originales (`df_clientes`, `df_ventas`, `df_marketing`) **no se modifican**. Trabajamos sobre copias (`*_clean`) para preservar la trazabilidad del proceso.

### 8.1 — Limpieza de `df_ventas`

```python
# 8.1.1 — Crear copia de trabajo

df_ventas_clean = df_ventas.copy()
print(f'Estado inicial: {df_ventas_clean.shape[0]} filas')
```

```python
# 8.1.2 — Eliminar duplicados completos

# .drop_duplicates() con keep='first' (default) elimina las repeticiones,
# manteniendo la primera ocurrencia. .reset_index(drop=True) reordena los
# índices para que sean consecutivos después del filtrado.
filas_antes = len(df_ventas_clean)
df_ventas_clean = df_ventas_clean.drop_duplicates(keep='first').reset_index(drop=True)
filas_despues = len(df_ventas_clean)

print(f'Duplicados eliminados: {filas_antes - filas_despues}')
print(f'Filas restantes: {filas_despues}')
```

```python
# 8.1.3 — Eliminar filas con nulos

# Antes de eliminar, verificamos qué filas se van a perder.
filas_con_nulos = df_ventas_clean[df_ventas_clean.isnull().any(axis=1)]
print(f'Filas con nulos a eliminar: {len(filas_con_nulos)}')
print(filas_con_nulos)
```

```python
# 8.1.4 — Aplicar dropna

# .dropna() sin parámetros elimina cualquier fila que tenga al menos un nulo.
# Como verificamos arriba, solo afecta a 2 filas con nulos en precio Y cantidad.
df_ventas_clean = df_ventas_clean.dropna().reset_index(drop=True)
print(f'Filas después de dropna: {len(df_ventas_clean)}')
print(f'Total de nulos restantes: {df_ventas_clean.isnull().sum().sum()}')
```

```python
# 8.1.5 — Limpiar la columna 'precio': sacar '$' y convertir a float

# Paso a paso:
#   .str.replace('$', '', regex=False) → reemplaza '$' por '' tratándolo como literal
#   .astype(float)                     → convierte el string ya limpio a número
df_ventas_clean['precio'] = (
    df_ventas_clean['precio']
    .str.replace('$', '', regex=False)
    .astype(float)
)

print(f'Tipo de precio: {df_ventas_clean["precio"].dtype}')
print(f'Estadísticos de precio:\n{df_ventas_clean["precio"].describe()}')
```

```python
# 8.1.6 — Convertir fecha_venta a datetime

df_ventas_clean['fecha_venta'] = pd.to_datetime(
    df_ventas_clean['fecha_venta'],
    dayfirst=True,        # las fechas vienen como dd/mm/yyyy
    errors='coerce'       # si hubiera fechas inválidas, las pone como NaT
)

print(f'Tipo de fecha_venta: {df_ventas_clean["fecha_venta"].dtype}')
print(f'Rango de fechas: {df_ventas_clean["fecha_venta"].min()} a {df_ventas_clean["fecha_venta"].max()}')
print(f'Fechas que no se pudieron convertir (NaT): {df_ventas_clean["fecha_venta"].isnull().sum()}')
```

## 9. Actividad 2.2 — Transformación de datos (Clase 6)

**Qué se hace:** aplicar filtros, selecciones y crear columnas derivadas para enfocar el análisis.

**Consigna específica:** "*Aplicar filtros y transformaciones para crear una tabla de ventas que muestre solo los productos con alto rendimiento.*"

**Definición operativa de "alto rendimiento":** un producto se considera de alto rendimiento si su **ingreso total** (precio × cantidad sumado en todo el período) está por encima del **percentil 75** del conjunto de productos. Es decir, el 25 % superior.

**Por qué este criterio:**
- Es **objetivo y reproducible** (no depende de un umbral arbitrario tipo "más de $10.000").
- Funciona independientemente de la escala monetaria del dataset.
- Es el criterio estándar para identificar el "top performers" en análisis comercial.

### 9.1 — Crear columna derivada `importe`

```python
# 9.1 — La columna 'importe' = precio * cantidad
# Esto es la primera transformación analítica útil del dataset limpio.

df_ventas_clean['importe'] = df_ventas_clean['precio'] * df_ventas_clean['cantidad']

print(f'Importe total facturado: $ {df_ventas_clean["importe"].sum():,.2f}')
print(f'\nEstadísticos de importe por venta:')
print(df_ventas_clean['importe'].describe())
```

### 9.2 — Identificar productos de alto rendimiento

```python
# 9.2.1 — Calcular ingresos totales por producto

ingresos_por_producto = (
    df_ventas_clean
    .groupby('producto')['importe']
    .sum()
    .sort_values(ascending=False)
)

print(f'Total de productos: {len(ingresos_por_producto)}')
print(f'\nTop 5 productos por ingresos:')
print(ingresos_por_producto.head().apply(lambda x: f'$ {x:>12,.2f}'))
print(f'\nBottom 5 productos por ingresos:')
print(ingresos_por_producto.tail().apply(lambda x: f'$ {x:>12,.2f}'))
```

```python
# 9.2.2 — Calcular el umbral de alto rendimiento (percentil 75)

umbral_p75 = ingresos_por_producto.quantile(0.75)
print(f'Umbral del percentil 75: $ {umbral_p75:,.2f}')

# Lista de productos cuyo ingreso supera el umbral
productos_alto_rendimiento = ingresos_por_producto[ingresos_por_producto >= umbral_p75].index.tolist()
print(f'\nProductos de alto rendimiento ({len(productos_alto_rendimiento)}):')
for p in productos_alto_rendimiento:
    print(f'  - {p:<25}  $ {ingresos_por_producto[p]:>12,.2f}')
```

### 9.3 — Filtrar `df_ventas_clean` a solo productos de alto rendimiento

```python
# 9.3 — Filtrado con boolean indexing

# .isin() devuelve True para las filas cuyo producto está en la lista de alto rendimiento.
mask = df_ventas_clean['producto'].isin(productos_alto_rendimiento)
df_ventas_top = df_ventas_clean[mask].reset_index(drop=True)

print(f'Filas en df_ventas_clean: {len(df_ventas_clean)}')
print(f'Filas en df_ventas_top:   {len(df_ventas_top)} ({len(df_ventas_top)/len(df_ventas_clean)*100:.1f}%)')
print(f'\nProductos únicos en df_ventas_top: {df_ventas_top["producto"].nunique()}')
print(f'Categorías representadas: {df_ventas_top["categoria"].unique().tolist()}')

df_ventas_top.head()
```

### 9.4 — Selección de columnas relevantes para el análisis

```python
# 9.4 — Selección de columnas

# Para los análisis posteriores no necesitamos id_venta. Construimos una vista
# con las columnas que aportan al análisis comercial.
df_ventas_top_view = df_ventas_top[[
    'producto', 'categoria', 'precio', 'cantidad', 'importe', 'fecha_venta'
]].copy()

print('Vista de las columnas relevantes:')
df_ventas_top_view.head(10)
```

### 9.5 — Transformaciones adicionales: columnas temporales

```python
# 9.5 — Crear columnas año, mes y día de la semana

# .dt es el accessor para columnas datetime, análogo a .str para strings.
df_ventas_clean['anio']            = df_ventas_clean['fecha_venta'].dt.year
df_ventas_clean['mes']             = df_ventas_clean['fecha_venta'].dt.month
df_ventas_clean['mes_nombre']      = df_ventas_clean['fecha_venta'].dt.month_name()
df_ventas_clean['dia_semana']      = df_ventas_clean['fecha_venta'].dt.day_name()
df_ventas_clean['periodo_mensual'] = df_ventas_clean['fecha_venta'].dt.to_period('M')

print('Columnas derivadas agregadas a df_ventas_clean:')
print(df_ventas_clean[['fecha_venta', 'anio', 'mes', 'mes_nombre', 'dia_semana']].head())
```

## 10. Actividad 2.3 — Agregación: ventas por categoría e ingresos (Clase 7)

**Qué se hace:** resumir el dataset limpio en métricas de negocio mediante `groupby()` y `pivot_table()`.

**Por qué se hace así:**
- `groupby` es la herramienta estándar para responder preguntas tipo "¿cuánto se vendió por X?".
- `pivot_table` es óptima para crear matrices comparativas (filas: una dimensión; columnas: otra).
- Ambas son traducciones directas de las consultas SQL `GROUP BY`.

**Preguntas de negocio que vamos a responder:**
1. ¿Qué categoría de producto genera más ingresos?
2. ¿Cuál es el comportamiento mensual de cada categoría?
3. ¿Cuáles son los productos más rentables dentro de cada categoría?
4. ¿Cómo se distribuye la inversión en marketing entre canales?

### 10.1 — Resumen de ventas por categoría

```python
# 10.1 — Múltiples métricas por categoría usando .agg()

# .agg() permite aplicar varias funciones de agregación en una sola operación.
# Esto es más eficiente y legible que hacer múltiples groupby.
ventas_por_categoria = df_ventas_clean.groupby('categoria').agg(
    cantidad_ventas=('id_venta',  'count'),     # cuántas transacciones
    unidades_vendidas=('cantidad', 'sum'),      # cuántas unidades en total
    precio_promedio=('precio',   'mean'),       # ticket promedio por unidad
    ingresos_totales=('importe', 'sum'),        # facturación total
).round(2)

# Agregamos el % sobre el total para tener perspectiva.
ventas_por_categoria['% del total'] = (
    ventas_por_categoria['ingresos_totales']
    / ventas_por_categoria['ingresos_totales'].sum() * 100
).round(2)

# Ordenamos por ingresos descendentes.
ventas_por_categoria = ventas_por_categoria.sort_values('ingresos_totales', ascending=False)
ventas_por_categoria
```

### 10.2 — Tabla dinámica: ventas por categoría y mes

```python
df_ventas_clean['periodo_mensual'] = df_ventas_clean['fecha_venta'].dt.to_period('M')

# 10.2 — pivot_table: filas=categoría, columnas=mes, valores=ingresos

# pivot_table es ideal para visualizar evolución temporal por categoría.
# margins=True agrega filas/columnas de totales (Total general).
pivot_categoria_mes = pd.pivot_table(
    df_ventas_clean,
    index='categoria',
    columns='periodo_mensual',
    values='importe',
    aggfunc='sum',
    fill_value=0,
    margins=True,
    margins_name='Total'
).round(2)

print('Ingresos por categoría y mes:')
pivot_categoria_mes
```

### 10.3 — Top productos dentro de cada categoría

```python
# 10.3 — Productos ordenados por ingresos dentro de cada categoría

# Agrupamos por dos niveles: categoría y producto.
ingresos_cat_prod = (
    df_ventas_clean
    .groupby(['categoria', 'producto'])['importe']
    .sum()
    .reset_index()
    .sort_values(['categoria', 'importe'], ascending=[True, False])
)

# Para cada categoría, los 5 productos top
print('Top productos por categoría:')
print('-' * 60)
for cat in ingresos_cat_prod['categoria'].unique():
    print(f'\n{cat.upper()}')
    subset = ingresos_cat_prod[ingresos_cat_prod['categoria'] == cat].head(5)
    for _, row in subset.iterrows():
        print(f"  {row['producto']:<25}  $ {row['importe']:>12,.2f}")
```

### 10.4 — Inversión en marketing por canal

```python
# 10.4 — Resumen de inversión en marketing por canal

inversion_por_canal = df_marketing_clean.groupby('canal').agg(
    cantidad_campanhas=('id_campanha', 'count'),
    inversion_total=('costo', 'sum'),
    inversion_promedio=('costo', 'mean'),
    duracion_promedio_dias=('duracion_dias', 'mean'),
).round(2)

inversion_por_canal['% del total'] = (
    inversion_por_canal['inversion_total']
    / inversion_por_canal['inversion_total'].sum() * 100
).round(2)

inversion_por_canal = inversion_por_canal.sort_values('inversion_total', ascending=False)
inversion_por_canal
```

### 10.5 — Tendencia mensual global de ventas

```python
# 10.5 — Evolución mes a mes con métricas agregadas

tendencia_mensual = df_ventas_clean.groupby('periodo_mensual').agg(
    ventas=('id_venta', 'count'),
    unidades=('cantidad', 'sum'),
    ingresos=('importe', 'sum'),
    ticket_promedio=('importe', 'mean'),
).round(2)

# variación porcentual mes a mes
tendencia_mensual['var_% ingresos'] = tendencia_mensual['ingresos'].pct_change().mul(100).round(2)

print('Tendencia mensual del negocio:')
tendencia_mensual
```

## 11. Actividad 2.4 — Integración de datos (Clase 8)

**Qué se hace:** combinar los datasets de ventas y marketing para obtener una vista integrada que permita analizar la efectividad de las campañas.

**Por qué se hace así:**
- Los 30 productos del dataset de marketing coinciden 100 % con los del dataset de ventas (verificado en EDA). El cruce por la columna `producto` es completo.
- `df_clientes` no tiene clave de cruce con ventas (no hay `id_cliente` en ventas), por lo que se mantiene como tabla independiente para análisis demográfico.
- Vamos a usar **`pd.merge()`** porque nos permite controlar el tipo de unión (`how`) y la(s) columna(s) clave (`on`).

**Diseño del cruce:**

| Tabla izquierda | Tabla derecha | Clave | Tipo de join | Resultado esperado |
|---|---|---|---|---|
| Resumen de ventas por producto | Resumen de marketing por producto | `producto` | `inner` | Solo productos que están en ambas tablas (los 30) |

### 11.1 — Construcción de las tablas a integrar

```python
# 11.1.1 — Resumen de ventas por producto

ventas_x_producto = df_ventas_clean.groupby('producto').agg(
    cantidad_ventas=('id_venta', 'count'),
    unidades_vendidas=('cantidad', 'sum'),
    ingresos=('importe', 'sum'),
).round(2).reset_index()

print(f'ventas_x_producto: {ventas_x_producto.shape}')
ventas_x_producto.head()
```

```python
# 11.1.2 — Resumen de marketing por producto

# Un producto puede tener múltiples campañas (en distintos canales).
# Agregamos por producto: total invertido y cantidad de campañas.
marketing_x_producto = df_marketing_clean.groupby('producto').agg(
    cantidad_campanhas=('id_campanha', 'count'),
    inversion_total=('costo', 'sum'),
    canales_usados=('canal', lambda x: ', '.join(sorted(x.unique()))),
).round(2).reset_index()

print(f'marketing_x_producto: {marketing_x_producto.shape}')
marketing_x_producto.head()
```

### 11.2 — Integración con `pd.merge()`

```python
# 11.2 — Merge de ventas y marketing por producto

# how='inner' devuelve solo los productos presentes en ambas tablas.
# Como ya verificamos que los 30 productos coinciden, el resultado debería tener 30 filas.
df_ventas_marketing = pd.merge(
    ventas_x_producto,
    marketing_x_producto,
    on='producto',
    how='inner',
    validate='one_to_one'   # garantiza que no haya duplicación inesperada
)

print(f'df_ventas_marketing: {df_ventas_marketing.shape}')
print(f'\nColumnas: {list(df_ventas_marketing.columns)}')
df_ventas_marketing.head()
```

### 11.3 — Métricas de efectividad: ROI por producto

```python
# 11.3 — Calcular el ROI (Return on Investment) por producto

# ROI = (ingresos - inversion) / inversion * 100
# Es la métrica clásica para medir efectividad de marketing.
df_ventas_marketing['roi_%'] = (
    (df_ventas_marketing['ingresos'] - df_ventas_marketing['inversion_total'])
    / df_ventas_marketing['inversion_total'] * 100
).round(2)

# Costo unitario de marketing por venta
df_ventas_marketing['costo_mkt_por_venta'] = (
    df_ventas_marketing['inversion_total'] / df_ventas_marketing['cantidad_ventas']
).round(2)

# Ingresos por unidad invertida en marketing (cuántos $ vuelven por cada $ gastado)
df_ventas_marketing['ingreso_x_$_invertido'] = (
    df_ventas_marketing['ingresos'] / df_ventas_marketing['inversion_total']
).round(2)

# Top 10 productos por ROI
print('Top 10 productos por ROI:')
df_ventas_marketing.sort_values('roi_%', ascending=False).head(10)
```

### 11.4 — Integración con tabla de clientes (vista paralela)

```python
# 11.4 — Análisis demográfico: clientes por ciudad

# Como df_clientes no se vincula con ventas, lo analizamos en paralelo.
# Agregamos la información disponible y cruzamos con totales nacionales.
clientes_por_ciudad = df_clientes_clean.groupby('ciudad').agg(
    cantidad_clientes=('id_cliente', 'count'),
    edad_promedio=('edad', 'mean'),
    ingreso_promedio=('ingresos', 'mean'),
).round(2).sort_values('cantidad_clientes', ascending=False)

clientes_por_ciudad['% clientes'] = (
    clientes_por_ciudad['cantidad_clientes']
    / clientes_por_ciudad['cantidad_clientes'].sum() * 100
).round(2)

clientes_por_ciudad
```

### 11.5 — Tabla integrada final: vista 360°

```python
# 11.5 — Vista integrada para análisis estratégico

# Mostramos la tabla integrada ordenada por ingresos, con todas las métricas relevantes.
vista_360 = df_ventas_marketing[[
    'producto', 'cantidad_ventas', 'unidades_vendidas', 'ingresos',
    'cantidad_campanhas', 'inversion_total', 'canales_usados',
    'roi_%', 'ingreso_x_$_invertido'
]].sort_values('ingresos', ascending=False).reset_index(drop=True)

print(f'Vista integrada: {vista_360.shape[0]} productos x {vista_360.shape[1]} columnas')
vista_360
```
