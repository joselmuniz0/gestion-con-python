```python
# 4.1 — Cálculo de ventas mensuales con Python

# Datos sintéticos de ejemplo (formato similar al dataset real)
ventas_demo = [
    {'fecha': '02/01/2024', 'producto': 'Cuadro decorativo', 'precio': 69.94, 'cantidad': 5},
    {'fecha': '15/01/2024', 'producto': 'Lámpara de mesa',   'precio': 105.10, 'cantidad': 3},
    {'fecha': '03/02/2024', 'producto': 'Heladera',          'precio': 850.00, 'cantidad': 2},
    {'fecha': '20/02/2024', 'producto': 'Tablet',            'precio': 320.00, 'cantidad': 4},
    {'fecha': '10/03/2024', 'producto': 'Smartphone',        'precio': 600.00, 'cantidad': 6},
]

# Acumulador: clave = mes (YYYY-MM), valor = importe total facturado en ese mes
ventas_por_mes = {}

for registro in ventas_demo:
    # La fecha viene como 'dd/mm/yyyy'. Extraemos mes y año por slicing.
    dia, mes, anio = registro['fecha'].split('/')   # ['02', '01', '2024']
    clave_mes = f'{anio}-{mes}'                     # '2024-01' (formato ordenable)

    # Operador aritmético: precio unitario * cantidad = importe de la línea
    importe = registro['precio'] * registro['cantidad']

    # .get(clave, 0) evita KeyError la primera vez que aparece un mes nuevo
    ventas_por_mes[clave_mes] = ventas_por_mes.get(clave_mes, 0) + importe

# Imprimimos ordenado cronológicamente (sorted ordena las claves alfabéticamente,
# y como usamos formato YYYY-MM eso coincide con el orden cronológico).
print('Ventas mensuales (versión didáctica con Python):')
print('-' * 40)
for mes in sorted(ventas_por_mes):
    print(f'  {mes}:  $ {ventas_por_mes[mes]:>12,.2f}')
```

### 4.2 — Versión con Pandas (sobre los datos reales)

Resolvemos el mismo cálculo sobre `df_ventas` usando vectorización. Es órdenes de magnitud más eficiente y es lo que se haría en un entorno profesional.

**Pasos:**
1. Crear una **copia local** del DataFrame (no tocamos el original durante el diagnóstico).
2. Limpiar localmente la columna `precio` para poder operar numéricamente con ella.
3. Convertir `fecha_venta` a `datetime`.
4. Calcular el importe línea por línea.
5. Agrupar por mes y sumar.

```python
# 4.2 — Cálculo de ventas mensuales con Pandas

# 1) Copia local: cualquier transformación que hagamos acá NO afecta a df_ventas.
#    Esto es clave en la Etapa 1: estamos diagnosticando, no limpiando.
ventas_calc = df_ventas.copy()

# 2) Limpieza local de la columna 'precio':
#    - .str.replace('$', '', regex=False) saca el símbolo monetario
#    - .astype(float) convierte el resultado a número
#    - errors='coerce' en pd.to_numeric: si algo no se puede convertir, queda NaN
ventas_calc['precio'] = pd.to_numeric(
    ventas_calc['precio'].str.replace('$', '', regex=False),
    errors='coerce'
)

# 3) Conversión de fecha: 'dd/mm/yyyy' -> datetime
#    dayfirst=True le indica a pandas el orden día-mes-año típico de Argentina
ventas_calc['fecha_venta'] = pd.to_datetime(
    ventas_calc['fecha_venta'], dayfirst=True, errors='coerce'
)

# 4) Cálculo del importe: precio * cantidad. Pandas vectoriza esto automáticamente.
ventas_calc['importe'] = ventas_calc['precio'] * ventas_calc['cantidad']

# 5) Agrupamiento por mes:
#    .dt.to_period('M') convierte cada fecha en un período mensual ('2024-01').
#    A diferencia de strftime(), preserva el orden cronológico nativo.
ventas_mensuales = (
    ventas_calc
    .groupby(ventas_calc['fecha_venta'].dt.to_period('M'))['importe']
    .sum()
    .sort_index()
)

print('Ventas mensuales reales (Pandas):')
print('-' * 40)
print(ventas_mensuales.apply(lambda x: f'$ {x:>14,.2f}').to_string())
print('-' * 40)
print(f'TOTAL del período: $ {ventas_mensuales.sum():,.2f}')
```

**Lectura del resultado:** el dataset cubre el período de febrero a octubre de 2024. Las dos filas con nulos en `precio` o `cantidad` se excluyen automáticamente del cálculo (al ser `NaN`, el producto da `NaN` y `sum()` lo ignora).

## 5. Actividad 3 — Estructuras de datos para almacenar ventas

Modelado de registros de ventas (producto, precio, cantidad) y argumentacion de la elección de estructura.

### Análisis de alternativas

| Modelo | Registro | Pros | Contras |
|---|---|---|---|
| Lista de listas | `['Heladera', 850.0, 2]` | Mínimo overhead, simple | Acceso por índice numérico (frágil si cambia el orden) |
| Lista de tuplas | `('Heladera', 850.0, 2)` | Inmutable: protege la integridad de cada venta | Sigue siendo posicional, no por nombre |
| **Lista de diccionarios** | `{'producto':'Heladera', 'precio':850.0, 'cantidad':2}` | **Acceso por nombre, autodescriptivo, convertible directo a DataFrame** | Más uso de memoria |
| Diccionario por producto | `{'Heladera': {'precio':850, 'unid':2}}` | Búsqueda O(1) por producto | Pierde detalle de ventas individuales (colapsa los duplicados) |

### Decisión: lista de diccionarios

**Razones técnicas:**
1. Cada venta tiene **campos heterogéneos con nombre** → el diccionario es la representación natural.
2. La **lista preserva el orden cronológico** de las ventas y admite duplicados legítimos (dos ventas del mismo producto el mismo día son válidas y no deben colapsarse).
3. Es **directamente convertible a `pandas.DataFrame`** con `pd.DataFrame(lista)` — esto cierra el ciclo con el resto del análisis.
4. Cuando se lee `v['precio']` se entiende inmediatamente qué se está accediendo, mientras que `v[1]` requiere recordar el orden de las columnas.

```python
# 5 — Almacenamiento de ventas con lista de diccionarios

# Subconjunto representativo del dataset real (productos de las 3 categorías)
ventas = [
    {'producto': 'Cuadro decorativo', 'precio':  69.94, 'cantidad': 5},
    {'producto': 'Lámpara de mesa',   'precio': 105.10, 'cantidad': 3},
    {'producto': 'Heladera',          'precio': 850.00, 'cantidad': 2},
    {'producto': 'Tablet',            'precio': 320.00, 'cantidad': 4},
    {'producto': 'Smartphone',        'precio': 600.00, 'cantidad': 6},
    {'producto': 'Heladera',          'precio': 850.00, 'cantidad': 1},  # duplicado legítimo
]

# 1) Iteración explícita: imprimir cada venta con su importe
print('Detalle línea por línea:')
print('-' * 60)
for v in ventas:
    importe = v['precio'] * v['cantidad']
    print(f"  {v['producto']:<20}  ${v['precio']:>7.2f} x {v['cantidad']}  =  ${importe:>10,.2f}")

# 2) Total general usando una generator expression dentro de sum()
#    Esto es más eficiente que crear una lista intermedia con [v['precio']*... for v in ventas]
total = sum(v['precio'] * v['cantidad'] for v in ventas)
print('-' * 60)
print(f'Total facturado: $ {total:,.2f}\n')

# 3) Total por producto: patrón acumulador con diccionario
totales_por_producto = {}
for v in ventas:
    p = v['producto']
    totales_por_producto[p] = totales_por_producto.get(p, 0) + v['precio'] * v['cantidad']

print('Total por producto:')
print('-' * 40)
for p, t in sorted(totales_por_producto.items(), key=lambda x: -x[1]):  # ordenado desc
    print(f'  {p:<20}  $ {t:>10,.2f}')

# 4) Conversión inmediata a DataFrame: puente con el resto del análisis
df_ventas_demo = pd.DataFrame(ventas)
print('\nMismo conjunto como DataFrame:')
df_ventas_demo
```

**Qué se hace:** describir cada dataset *antes* de cualquier transformación, respondiendo:

- ¿Qué tipo tiene cada columna?

**Por qué se hace así:** sin entender el dato, cualquier limpieza posterior sería a ciegas. El EDA es el insumo que nos permite decidir, en la Etapa 2, **qué** hay que limpiar y **cómo**.

Encapsulamos la exploración en una función reutilizable para garantizar que los 3 datasets se traten con el mismo criterio sistemático.

```python
# 6.1 — Función reutilizable de EDA

def explorar_dataframe(df: pd.DataFrame, nombre: str) -> None:
    '''
    Imprime un resumen exploratorio estándar de un DataFrame.

    Parameters
    ----------
    df : pd.DataFrame
        DataFrame a explorar.
    nombre : str
        Nombre descriptivo del dataset (aparece en los encabezados).
    '''
    sep = '=' * 70
    print(sep)
    print(f'  EDA — {nombre}')
    print(sep)

    # 1. Dimensiones del dataset
    print(f'\nForma: {df.shape[0]} filas x {df.shape[1]} columnas')

    # 2. Tipos de datos y uso de memoria
    print('\nTipos de datos y memoria:')
    df.info(memory_usage='deep')

    # 3. Primeras filas
    print('\nPrimeras 5 filas:')
    print(df.head())

    # 4. Últimas filas (útil para detectar registros incompletos al final)
    print('\nÚltimas 5 filas:')
    print(df.tail())

    # 5. Estadísticos descriptivos de variables numéricas
    print('\nEstadísticos numéricos:')
    print(df.describe())

    # 6. Estadísticos de variables categóricas (si existen)
    cols_obj = df.select_dtypes(include=['object', 'category']).columns
    if len(cols_obj) > 0:
        print('\nEstadísticos de columnas categóricas:')
        print(df[cols_obj].describe())

    # 7. Cardinalidad: cuántos valores únicos tiene cada columna
    print('\nValores únicos por columna:')
    print(df.nunique().sort_values(ascending=False))

    print('\n' + sep + '\n')
```

## 7. Actividad 5 — Calidad de datos: nulos y duplicados

Se identifican valores nulos y registros duplicados, se emite un informe consolidado del estado inicial.



```python
# 7.1 — Función de informe de nulos por columna (con porcentaje)

def informe_nulos(df: pd.DataFrame, nombre: str) -> pd.DataFrame:
    '''
    Devuelve un DataFrame con cantidad y porcentaje de nulos por columna.

    Reportar cantidad y porcentaje permite priorizar el tratamiento:
    - <5% de nulos → suele admitir imputación o eliminación de filas.
    - 5–30%       → analizar con cuidado el impacto.
    - >30%        → considerar descartar la columna.
    '''
    cantidad   = df.isnull().sum()
    porcentaje = (cantidad / len(df) * 100).round(2)
    informe = (
        pd.DataFrame({
            'columna': cantidad.index,
            'nulos': cantidad.values,
            '% nulos': porcentaje.values,
        })
        .sort_values('nulos', ascending=False)
        .reset_index(drop=True)
    )
    print(f'\n--- Nulos en {nombre} ---')
    print(f'Total de nulos en el dataset: {int(df.isnull().sum().sum())}')
    print(informe.to_string(index=False))
    return informe

nulos_clientes  = informe_nulos(df_clientes,  'Clientes')
nulos_ventas    = informe_nulos(df_ventas,    'Ventas')
nulos_marketing = informe_nulos(df_marketing, 'Marketing')
```

```python
# 7.2 — Función de informe de duplicados

def informe_duplicados(df: pd.DataFrame, nombre: str, subset=None) -> pd.DataFrame:
    '''
    Reporta y devuelve los registros duplicados de un DataFrame.

    Parameters
    ----------
    subset : list[str] | None
        Columnas que definen 'duplicado'. Si es None, considera todas.

    Notas:
    - keep=False marca *todas* las apariciones del duplicado, no solo las repeticiones.
    - keep='first' (default) marcaría solo las repeticiones, ocultando la primera ocurrencia.
    '''
    mask = df.duplicated(subset=subset, keep=False)
    duplicados = df[mask].sort_values(by=subset or list(df.columns))

    print(f'\n--- Duplicados en {nombre} ---')
    print(f'Filas totales: {len(df)}')
    print(f'Filas duplicadas (todas las apariciones): {len(duplicados)}')
    print(f'Filas únicas si se eliminaran duplicados: '
          f'{len(df) - int(df.duplicated(subset=subset).sum())}')
    if subset:
        print(f'Criterio de duplicación: columnas {subset}')
    return duplicados

dup_clientes  = informe_duplicados(df_clientes,  'Clientes')
dup_ventas    = informe_duplicados(df_ventas,    'Ventas')
dup_marketing = informe_duplicados(df_marketing, 'Marketing')
```

```python
# 7.3 — Informe consolidado del estado inicial

def informe_inicial(df: pd.DataFrame, nombre: str) -> dict:
    '''Imprime y devuelve el informe inicial de calidad de un dataset.'''
    print(f'\n{"#"*60}')
    print(f'  INFORME INICIAL — {nombre}')
    print(f'{"#"*60}')

    total_registros = len(df)
    nulos_por_col   = df.isnull().sum()
    total_nulos     = int(nulos_por_col.sum())
    duplicados      = int(df.duplicated().sum())

    print(f'\na) Cantidad total de registros: {total_registros}')
    print(f'\nb) Valores nulos por columna:')
    if total_nulos > 0:
        print(nulos_por_col[nulos_por_col > 0].to_string())
    else:
        print('   (no hay nulos)')
    print(f'\n   Total de nulos en el dataset: {total_nulos}')
    print(f'\nc) Filas duplicadas (criterio: todas las columnas): {duplicados}')

    return {
        'dataset': nombre,
        'registros': total_registros,
        'nulos_total': total_nulos,
        'nulos_por_columna': nulos_por_col.to_dict(),
        'duplicados_total': duplicados,
    }

resumen_clientes  = informe_inicial(df_clientes,  'Clientes')
resumen_ventas    = informe_inicial(df_ventas,    'Ventas')
resumen_marketing = informe_inicial(df_marketing, 'Marketing')
```
